# Bài 1: Mô hình phân lớp trên bộ dữ liệu Iris (thư viện sklearn)

Bộ dữ liệu Iris gồm thông tin về 3 loài hoa Iris: **Iris setosa**, **Iris versicolor** và **Iris virginica**.

Gồm 4 thuộc tính đo lường (đơn vị cm): chiều dài lá đài (sepal length), chiều rộng lá đài (sepal width), chiều dài cánh hoa (petal length), chiều rộng cánh hoa (petal width).

## a) Đọc bộ dữ liệu Iris từ thư viện sklearn

*Hướng dẫn: Từ thư viện sklearn import Iris và sử dụng hàm `load_iris` để đọc bộ dữ liệu.*

```python
print("Họ và tên: Lê Quốc Trung - MSSV: 6451071082")
import pandas as pd
from sklearn.datasets import load_iris

# Đọc bộ dữ liệu Iris
iris = load_iris()
print(type(iris))
print(iris.keys())
```

## b) Khảo sát bộ dữ liệu Iris

### - In ra tên các thuộc tính, thuộc tính phân lớp

*Hướng dẫn: Sử dụng `feature_names`, `target_names`.*

```python
print("Họ và tên: Lê Quốc Trung - MSSV: 6451071082")
print("Tên các thuộc tính (feature_names):", iris.feature_names)
print("Tên các lớp (target_names):", iris.target_names)
```

### - In ra số lượng mẫu, số lượng thuộc tính

*Hướng dẫn: Sử dụng `data.shape`, `target.shape`.*

```python
print("Họ và tên: Lê Quốc Trung - MSSV: 6451071082")
print("Kích thước data (mẫu, thuộc tính):", iris.data.shape)
print("Kích thước target:", iris.target.shape)
print("Số lượng mẫu:", iris.data.shape[0])
print("Số lượng thuộc tính:", iris.data.shape[1])
```

### - In ra 5 mẫu dữ liệu đầu tiên

*Hướng dẫn: Sử dụng `data[]`, `target[]`.*

```python
print("Họ và tên: Lê Quốc Trung - MSSV: 6451071082")
print("5 mẫu dữ liệu đầu tiên (data):")
print(iris.data[:5])
print("\nNhãn tương ứng (target):")
print(iris.target[:5])
```

### - Thống kê count, mean, std, min, max, tứ phân vị của bộ dữ liệu

*Hướng dẫn: Tạo dataframe và sử dụng `describe`.*

```python
print("Họ và tên: Lê Quốc Trung - MSSV: 6451071082")
# Tạo dataframe từ dữ liệu Iris
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target

# Thống kê count, mean, std, min, 25%, 50%, 75%, max
df.describe()
```
