# Bài 2: Mô hình phân lớp trên bộ dữ liệu MNIST (thư viện Keras)

Bộ dữ liệu MNIST được tạo ra từ một tập hợp các ảnh kỹ thuật số của các chữ số viết tay từ 0 đến 9. Gồm 70.000 ảnh, trong đó 60.000 ảnh được sử dụng cho train và 10.000 ảnh được sử dụng cho test.

Mỗi ảnh có kích thước 28x28 pixel, là ảnh đen trắng (grayscale). Mỗi pixel là giá trị từ 0 đến 255 (0 là đen tuyệt đối, 255 là trắng tuyệt đối). Nhãn là số nguyên từ 0 đến 9.

## a) Đọc dữ liệu train, test của MNIST từ Keras

*Hướng dẫn: Từ thư viện Keras import MNIST và sử dụng hàm `load_data` để đọc bộ dữ liệu và ghi vào `(train_images, train_labels), (test_images, test_labels)`.*

```python
print("Họ và tên: Lê Quốc Trung - MSSV: 6451071082")
from tensorflow.keras.datasets import mnist

# Đọc dữ liệu train, test
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
print("Đã đọc xong bộ dữ liệu MNIST")
```

## b) In ra kích thước của tập train, test

*Hướng dẫn: Sử dụng `shape`.*

```python
print("Họ và tên: Lê Quốc Trung - MSSV: 6451071082")
print("Kích thước train_images:", train_images.shape)
print("Kích thước train_labels:", train_labels.shape)
print("Kích thước test_images:", test_images.shape)
print("Kích thước test_labels:", test_labels.shape)
```
